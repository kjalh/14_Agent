```python
def solution(s):
    answer = True
    
    if len(s)>100000:
        return False
    
    s = list(s)
    if s[0] != "(" or s[-1] != ")":
        return False

    qq = []
    pp = []
    
    for i in s:
        if i == "(":
            qq.append(i)
        elif i  == ")":
            pp.append(i)
        
        s.pop() 
    
    print(qq)
    print(pp)
    print(len(qq))
    print(len(pp))
    
    if len(qq) == len([pp]):
        return True
    else:
        return False

solution("(())()")
```
